```python
# To Check Whether Or Not The Triangle Is A Right-Angled Triangle Using Function

def check_right_angled_triangle(a,b,c):
    sides = sorted([a,b,c])
    if sides[0] + sides[1] <= sides[2]:
        return "Triangle cannot be formed."
    if sides[0]**2 + sides[1]**2 == sides[2]**2:
        return "Given sides make a right-angled triangle."
    else:
        return "Given sides don't make a right-angled triangle."
x = float(input("Enter 1st side:"))
y = float(input("Enter 2nd side:"))
z = float(input("Enter 3rd side:"))
print(check_right_angled_triangle(x,y,z))
```
