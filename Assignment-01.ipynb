```python
# Different Operations On Dictionary & Tuple Data Structure

Student_List=["Prachi","Shruti","Nandini","Shanvi","Swanandi"]
Student_Tuple=("Shweta","Ranbir","Kajal","Sameer","Priya")
Student_Dict={
    101:"Prachi",
    102:"Shruti",
    103:"Nandini",
    104:"Shanvi",
    105:"Swanandi"
}
print(f"Student List:{Student_List}")
print(f"Student Tuple:{Student_Tuple}")
print(f"Student Dictionary:{Student_Dict}")
Student_Dict[106]="Trisha"
print(f"Student Dictionary After Adding Data 106:{Student_Dict}")
Student_Dict[101]="Ayushi"
print(f"Student Dictionary After Updating Data 101:{Student_Dict}")
del Student_Dict[102]
print(f"Student Dictionary After Deleting Data 102:{Student_Dict}")
```
