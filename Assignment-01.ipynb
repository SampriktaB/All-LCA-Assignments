```python
# Student Data Management Using Python Collections (Dictionary, Tuple & List)

Students = {
    101: ("Nandini","CSE",[76, 87, 93]),
    102: ("Shruti","CSE(AI & DS)",[81, 95, 75]),
    103: ("Nipun","ECE",[84, 92, 98])
}
print("Original Student Records:")
print(Students)

# Adding A New Student Record 
Students[104] = ("Ayushi","CSE",[83, 90, 72])
print("\nAfter Adding A New Student:")
print(Students)

# Deleting An Existing Student Record 
del Students[101]
print("\nAfter Deleting Student 101:")
print(Students)

# Updating Details Of A Student 
Students[102] = ("Trisha","CSE(AI & DS)",[94, 85, 88])
print("\nAfter Updating Student 102:")
print(Students)

# Displaying Final Records 
print("\nFinal Student Records:")
for Roll_Number in Students:
    print("\nRoll Number:", Roll_Number)
    print("Name:", Students[Roll_Number][0])
    print("Branch:", Students[Roll_Number][1])
    print ("Marks:", Students[Roll_Number][2])
```
